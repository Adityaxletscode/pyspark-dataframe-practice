#***Employee Salary Filter***

You are given an employee dataset containing:

employee_id, employee_name, department, salary, city

The HR team wants to identify employees who are earning more than ₹50,000 and working in the IT department.

Task: Create a DataFrame, filter the required employees, and display only employee_id, employee_name, department, and salary. Sort the result by salary from highest to lowest.

Concepts: createDataFrame(), filter(), select(), orderBy()

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrame Assignment").getOrCreate()
```

```python
df = spark.read.csv("/Volumes/aditya_catalog/aditya_schema/aditya_volume/employee_dataset.csv", header=True, inferSchema=True)
df.show()
df.printSchema()
```

```python
result_df=df.filter((df.salary>50000)&(df.department=='IT'))
result_df.select("employee_id","employee_name","department","salary").orderBy("salary",ascending=False).show()
```

#***Categorize Customers Based on Age***

A retail company wants to classify its customers into different age groups. The customer dataset contains:

customer_id, customer_name, age, city

Create a new column called age_category using the following conditions:

Age < 25        → Young
Age 25 to 40    → Adult
Age > 40        → Senior

Task: Display the customer name, age, city, and calculated age category.

Concepts: withColumn(), when(), otherwise()

```python
cust_df = spark.read.csv("/Volumes/aditya_catalog/aditya_schema/aditya_volume/customer_age_groups.csv", header=True, inferSchema=True)
cust_df.show()
cust_df.printSchema()
```

```python
from pyspark.sql.functions import *
new_cust_df=cust_df.withColumn("age_group",when(col("age")<25,"Young").when((col("age")>=25) & (col("age")<=40),"Adult").otherwise("Senior"))
new_cust_df.select("customer_name","age","city","age_group").show()
```

#***Completed Order Sales Analysis***

An e-commerce company maintains order data containing:

order_id, customer_id, product, category, quantity, unit_price, status

The business team wants to analyze only successfully completed orders.

Task: Filter records where status = "Completed". Create a new column:

order_value = quantity × unit_price

Then calculate the total sales amount for each category and display categories from highest to lowest sales.

Concepts: filter(), withColumn(), groupBy(), sum(), orderBy()

```python
ecom_df = spark.read.csv("/Volumes/aditya_catalog/aditya_schema/aditya_volume/ecommerce_orders.csv", header=True, inferSchema=True)
ecom_df.show()
ecom_df.printSchema()
```

```python
suscom_df=ecom_df.filter(col("status")=='Completed')
suscom_df.show()
```

```python
new_df=suscom_df.withColumn("order_value",col("quantity")*col("unit_price"))
new_df.show()
```

```python
result=new_df.groupby("category").agg(sum("order_value").alias("Total sales")).orderBy("Total sales",ascending=False)
result.show()
```

#***Find Customers Without Orders***

The HR department wants to identify the highest-paid employee in every department.

The dataset contains:

employee_id, employee_name, department, salary

Task: Create a window for each department, arrange employees by salary in descending order, assign a row number, and return only the highest-paid employee from each department.

Expected output:

department | employee_name | salary
IT         | Karthik       | 80000
HR         | Priya         | 65000
Finance    | John          | 75000

Concepts: Window.partitionBy(), orderBy(), row_number(), filter()


```python
df.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *

wind=Window.partitionBy("department").orderBy(col("salary").desc())
wind_df=df.withColumn("rank",row_number().over(wind))
result=wind_df.filter(col("rank")==1).show()
```

#***E-Commerce Customer Spending Analysis***

An e-commerce company wants to identify its most valuable customers. You are provided with three DataFrames.

Customers

customer_id, customer_name, city

Orders

order_id, customer_id, order_date, status

Order Items

order_id, product_name, category, quantity, unit_price

Task: Perform the following operations:

Keep only orders where status = "Completed".
Calculate item_value = quantity × unit_price.
Calculate the total value of each order.
Join the order totals with the completed orders.
Join the result with the customer DataFrame.
Calculate the total spending of every customer.
Keep only customers whose total spending is greater than ₹10,000.
Sort customers by total spending in descending order.
Display the top 5 customers.
Produce the final result as:
customer_id | customer_name | city | total_spending

Concepts: filter(), withColumn(), groupBy(), agg(), join(), select(), orderBy(), limit()

```python
ord3_df = spark.read.csv("/Volumes/aditya_catalog/aditya_schema/aditya_volume/orders_training_dataset.csv", header=True, inferSchema=True)
item_df = spark.read.csv("/Volumes/aditya_catalog/aditya_schema/aditya_volume/order_items_training_dataset.csv", header=True, inferSchema=True)
cust_df.show()
ord3_df.show()
item_df.show()
```

```python
cust_df.show()
```

```python
ord3_df.show()
```

```python
item_df.show()
```

```python
orders=ord3_df.filter(col("status")=="Completed")
orders.show()
```

```python
items=item_df.withColumn("item_value",item_df.quantity*item_df.unit_price)
items.show()
```

```python
items2=items.groupBy("order_id").agg(sum("item_value").alias("amount"))
items2.show()
```

```python
orders_join=orders.join(items2,orders.order_id==items2.order_id).select("customer_id",orders.order_id,"order_date","amount")
orders_join.show()
```

```python
cust_join=cust_df.join(orders_join,cust_df.customer_id==orders_join.customer_id).select(cust_df.customer_id,"customer_name","city","order_id","amount")
cust_join.show()
```

```python
cust_res=cust_join.groupBy("customer_id","customer_name","city").agg(sum("amount").alias("Total_spending"))
cust_res.show()
```

```python
result=cust_res.filter(col("Total_spending")>10000).orderBy("Total_spending",ascending=False)
result.show()
```
